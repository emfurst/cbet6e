```python
import numpy as np
import matplotlib.pyplot as plt
```

The first step is to create numpy arrays of the data.

```python
# Data, T in C volume in cm^3

T1 = np.arange(0,10,1)
T2 = np.linspace(10,100,10)
T = np.concatenate((T1,T2))

Vh2o = np.array([1.0001329, 1.0000733, 1.0000321, 1.0000078, 1.0000000, 
                 1.0000081, 1.0000318, 1.0000704, 1.0001236, 1.0001909,
                 1.0002719, 1.0015678, 1.0043408, 1.0078108, 1.012074,
                 1.017046, 1.022694, 1.028987, 1.035904, 1.043427])

Vhg = np.array([0.0735560, 0.0735694, 0.0735828, 0.0735961, 0.0736095,
                0.0736228, 0.0736362, 0.0736496, 0.0736629, 0.0736763,
                0.0736893, 0.0738233, 0.0739572, 0.0740910, 0.0742250,
                0.0743592, 0.0744936, 0.0746282, 0.0747631, 0.0748981])
```

Now we can plot the data.

```python
# Plot the data

fig, ax = plt.subplots()
ax.plot(T, Vh2o, 'o')
ax.plot(T, Vhg, 'o')
plt.xlabel('T (C)')
plt.ylabel('Volume (cm^3)')
plt.title('Volume of water and mercury with temperature')
```

On a single plot, the specific volumes of water and mercury are too different from each another. Let's plot them separately.

## Plot 1: Water specifc volume with temperature

```python
fig, ax = plt.subplots()
ax.plot(T, Vh2o, 'o')
plt.xlabel('T (C)')
plt.ylabel('Volume (cm^3)')
plt.title('Volume of water with temperature')
```

## Plot  2: Mercury specific volume with temperature

```python
fig, ax = plt.subplots()
ax.plot(T, Vhg, 'o')
plt.xlabel('T (C)')
plt.ylabel('Volume (cm^3)')
plt.title('Volume of mercury with temperature')
```
