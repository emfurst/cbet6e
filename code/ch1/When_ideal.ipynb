# When is a gas ideal?

In this notebook we'll construct a PV diagram using the van der Waals (vdw) equation of state (EOS) and the ideal gas law. This will give us some beginning appreciation for when a gas can be considered ideal or not. 

The van der Waals EOS is:

$$ (P + a/\underline{V}^2)(\underline{V}-b) = RT$$

Remember, $ \underline{V} $ is the molar volume (m^3/mol).

The ideal gas law is

$$ P\underline{V} = RT $$

```python
import numpy as np
```

```python
"""
Constants for the vdW equation for CO2
"""

a = 0.3658    # Pa m^6/mol^2
b = 4.286e-5  # m^3/mol

"""
Other parameters
"""
R = 8.314   # J/mol.K  R = the ideal gas constant
T = 300     # K, temperature

"""
Pressure range to consider
"""
# We want a range of pressures from 1 atm to 100 atm
P_min = 1.01325e5    # Pa
P_max = 1.01325e7    # Pa
n_points = 100       # number of points in the pressure range 
P = np.linspace(P_min, P_max, n_points)
```

```python
V_ideal = R*T/P
print(V_ideal)
```

```python
import scipy.optimize as opt

"""
Constants for the vdW equation for CO2
"""

a = 0.3658    # Pa m^6/mol^2
b = 4.286e-5  # m^3/mol

"""
Other parameters
"""

R = 8.314   # J/mol.K  R = the ideal gas constant

def vdW(V):
    x = R*T - (P + a/V**2)*(V-b)
    return x

V_vdw = opt.fsolve(vdW, R*T/P)

print(V_vdw)
# has units of mol/m3
```
