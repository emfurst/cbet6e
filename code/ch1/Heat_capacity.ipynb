# Heat capacity

Using the empirical equation

$$C_p^*(T) = a + bT + cT^2 + dT^3$$

calculate the enthalpy and internal energy change of a gas in the ideal state.

October 2025

```python
R = 8.314 # J/(mol K)
```

```python
# Data for oxygen
a = 25.460
b = 1.519*10**-2
c = -0.715*10**-5
d = 1.311*10**-9

T1 = 300 # temperature in K
T2 = 500 # temperature in K
```

```python
# Cp = a + b*T + c*T**2 + d*T**3
# Integrating the funciton yields a change in the molar enthalpy:

delta_H = a*(T2 - T1) + (b/2)*(T2**2 - T1**2) + (c/3)*(T2**3 - T1**3) + (d/4)*(T2**4 - T1**4)
print(delta_H)
```

```python
# Propane
a = -4.042
b = 30.456*10**-2
c = -15.711*10**-5
d = 31.716*10**-9

T1 = 300
T2 = 500
```

```python
delta_H = a*(T2 - T1) + (b/2)*(T2**2 - T1**2) + (c/3)*(T2**3 - T1**3) + (d/4)*(T2**4 - T1**4)
print(delta_H)
```

```python
delta_U = (a-R)*(T2 - T1) + (b/2)*(T2**2 - T1**2) + (c/3)*(T2**3 - T1**3) + (d/4)*(T2**4 - T1**4)
print(delta_U)
```

```python
print(f"Constant pressure (5 mol): {delta_H*5/1000:.2g} kJ")
print(f"Constant volume (5 mol): {delta_U*5/1000:.2g} kJ")
```
